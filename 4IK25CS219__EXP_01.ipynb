```python
import numpy as np
import pandas as pd

a = np.array([37, 66, 78, 55, 46])

print('Array is', a)
print('Mean is', a.mean())
print('Median is', np.median(a))
print('Reshaped 2D:\n', a.reshape(5, 1))
print('Sliced array', a[1:4])

data = {
    'Name': ['Purvi', 'Tanmai', 'Vibha', 'Trupti'],
    'Marks': [55, 86, 37, 93],
    'Branch': ['CSE', 'ECE', 'ISE', 'AIML']
}

df = pd.DataFrame(data)

print('First two rows:\n', df.iloc[0:2])
print('Marks column:\n', df['Marks'])
print('Rows where marks greater than 75:\n', df[df['Marks'] > 75])
```
